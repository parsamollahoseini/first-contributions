```python
"""
Author: Your Real First and Last Name
Assignment: #1
"""

# Create 4 variables with their data types.
gym_member = "Alex Alliton"    # str
preferred_weight_kg = 20.5     # float
highest_reps = 25              # int
membership_active = True       # bool

# workout_stats is a dictionary with friend names as keys and tuples of workout minutes as values.
workout_stats = {
    "Alex": (30, 45, 20),
    "Jamie": (40, 30, 25),
    "Taylor": (20, 50, 30)
}

# Calculate and add total workout minutes for each friend.
for friend, workouts in list(workout_stats.items()):
    if isinstance(workouts, tuple):
        workout_stats[f"{friend}_Total"] = sum(workouts)

# workout_list is a 2D (nested) list where each row represents a friend and each column represents an activity.
workout_list = [list(workouts) for key, workouts in workout_stats.items() if not key.endswith("_Total")]

# Slice workout_list to extract and print yoga and running minutes for all friends.
yoga_running = [row[:2] for row in workout_list]
print("Yoga and Running minutes for all friends:")
for row in yoga_running:
    print(row)

# Slice workout_list to extract and print weightlifting minutes for the last two friends.
weightlifting_last_two = [row[2] for row in workout_list[-2:]]
print("\nWeightlifting minutes for the last two friends:")
for minutes in weightlifting_last_two:
    print(minutes)

# Check each friend's total workout minutes; if >= 120, print a congratulatory message.
for key in workout_stats:
    if key.endswith("_Total") and workout_stats[key] >= 120:
        friend_name = key.replace("_Total", "")
        print(f"\nGreat job staying active, {friend_name}!")

# Allow the user to input a friend's name and display their workout details if found.
user_friend = input("\nEnter a friend's name: ")
if user_friend in workout_stats and isinstance(workout_stats[user_friend], tuple):
    workouts = workout_stats[user_friend]
    total = workout_stats.get(f"{user_friend}_Total", sum(workouts))
    print(f"\n{user_friend}'s workout minutes:")
    print(f"  Yoga: {workouts[0]}")
    print(f"  Running: {workouts[1]}")
    print(f"  Weightlifting: {workouts[2]}")
    print(f"  Total: {total}")
else:
    print(f"\nFriend {user_friend} not found in the records.")

# Determine and print the friend with the highest and the lowest total workout minutes.
totals = {key.replace("_Total", ""): value for key, value in workout_stats.items() if key.endswith("_Total")}
if totals:
    highest_friend = max(totals, key=totals.get)
    lowest_friend = min(totals, key=totals.get)
    print(f"\nFriend with highest total workout minutes: {highest_friend} ({totals[highest_friend]})")
    print(f"Friend with lowest total workout minutes: {lowest_friend} ({totals[lowest_friend]})")
```
